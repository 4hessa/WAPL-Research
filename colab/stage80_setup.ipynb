## 2. حفظ رمز GitHub (مربع مخفي)

````python
import getpass, os
t = getpass.getpass('GitHub token: ').strip()
p = os.path.expanduser('~/.git-credentials')
open(p, 'w').write(f'https://x-access-token:{t}@github.com\n')
os.chmod(p, 0o600)
print('saved; token length =', len(t))
del t
````

## 3. تثبيت Claude Code وتنزيل المستودع وإعداد git

````python
!curl -fsSL https://claude.ai/install.sh | bash
!grep -q '.local/bin' ~/.bashrc || echo 'export PATH="$HOME/.local/bin:$PATH"' >> ~/.bashrc
%cd /content
![ -d WAPL-Research ] || git clone -q https://github.com/4hessa/WAPL-Research.git
%cd /content/WAPL-Research
!git pull -q origin main && git log --oneline -1
!git config --global user.name 'WAPL Colab' && git config --global user.email 'colab@users.noreply.github.com' && git config --global credential.helper store
!ls -l ~/.local/bin/claude && echo 'ready'
````
